```python
# ============================================================
# CELL 1: IMPORT LIBRARIES
# Mục đích: Import các thư viện cần thiết cho quá trình EDA
# ============================================================

# Pandas: Đọc và xử lý dữ liệu dạng bảng
import pandas as pd

# NumPy: Hỗ trợ tính toán số học
import numpy as np

# Matplotlib: Thư viện vẽ biểu đồ
import matplotlib.pyplot as plt

# Seaborn: Hỗ trợ trực quan hóa dữ liệu đẹp và dễ đọc hơn
import seaborn as sns

# Path: Hỗ trợ xử lý đường dẫn file và thư mục
from pathlib import Path

# Thiết lập giao diện mặc định cho biểu đồ
sns.set_theme(style="whitegrid")

# Thiết lập kích thước biểu đồ mặc định
plt.rcParams["figure.figsize"] = (8, 5)

# Hiển thị phiên bản các thư viện để kiểm tra môi trường
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)

print("Libraries imported successfully!")

# ============================================================
# CELL 2: LOAD DATASET
# Mục đích: Đọc dữ liệu nhân sự từ file CSV
# ============================================================

# Xác định thư mục gốc của dự án
# Giả sử Notebook đang nằm trong thư mục notebooks
PROJECT_DIR = Path.cwd().parent

# Xác định đường dẫn đến file CSV
DATA_PATH = (
    PROJECT_DIR
    / "dataset"
    / "WA_Fn-UseC_-HR-Employee-Attrition.csv"
)

# Kiểm tra file dataset có tồn tại hay không
if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Không tìm thấy dataset tại: {DATA_PATH}"
    )

# Đọc dữ liệu CSV vào DataFrame
df = pd.read_csv(DATA_PATH)

# Hiển thị thông báo khi đọc dữ liệu thành công
print("Dataset loaded successfully!")

# Hiển thị kích thước dữ liệu
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Hiển thị 5 dòng dữ liệu đầu tiên
display(df.head())


# ============================================================
# CELL 3: DATA OVERVIEW
# Mục đích: Tìm hiểu cấu trúc và đặc điểm tổng quan của dataset
# ============================================================

# Hiển thị kích thước dataset
print("Dataset shape:", df.shape)

# Hiển thị thông tin các cột:
# - Tên cột
# - Số lượng giá trị không bị thiếu
# - Kiểu dữ liệu
print("\nDataset information:")
df.info()

# Thống kê các cột dạng số:
# count: Số lượng giá trị
# mean: Giá trị trung bình
# std: Độ lệch chuẩn
# min/max: Giá trị nhỏ nhất/lớn nhất
print("\nNumerical statistics:")
display(df.describe())

# Thống kê tất cả các cột, bao gồm cả dữ liệu dạng chữ
print("\nComplete statistics:")
display(df.describe(include="all"))

# ============================================================
# CELL 4: FEATURE CLASSIFICATION
# Mục đích: Phân loại các biến thành biến số và biến phân loại
# ============================================================

# Xác định cột mục tiêu cần dự đoán
TARGET = "Attrition"

# Lấy danh sách các cột có kiểu dữ liệu object
# Ví dụ: Gender, Department, JobRole, OverTime
categorical_cols = df.select_dtypes(
    include=["object"]
).columns.tolist()

# Lấy danh sách các cột dạng số
# Ví dụ: Age, MonthlyIncome, YearsAtCompany
numerical_cols = df.select_dtypes(
    include=["number"]
).columns.tolist()

# Loại bỏ biến mục tiêu khỏi danh sách biến đầu vào
categorical_features = [
    col for col in categorical_cols
    if col != TARGET
]

numerical_features = [
    col for col in numerical_cols
    if col != TARGET
]

print("Target variable:", TARGET)

print("\nCategorical features:")
print(categorical_features)

print("\nNumerical features:")
print(numerical_features)

print("\nNumber of categorical features:", len(categorical_features))
print("Number of numerical features:", len(numerical_features))

# ============================================================
# CELL 5: ATTRITION DISTRIBUTION
# Mục đích: Phân tích tỷ lệ nhân viên nghỉ việc và ở lại
# ============================================================

# Đếm số lượng nhân viên theo từng trạng thái nghỉ việc
attrition_count = df[TARGET].value_counts()

# Tính tỷ lệ phần trăm của từng trạng thái
attrition_percentage = (
    df[TARGET].value_counts(normalize=True) * 100
)

# Hiển thị số lượng nhân viên
print("Employee Attrition Count:")
print(attrition_count)

# Hiển thị tỷ lệ phần trăm
print("\nEmployee Attrition Percentage:")
print(attrition_percentage.round(2))

# Tạo biểu đồ gồm hai phần:
# Bên trái: Số lượng nhân viên
# Bên phải: Tỷ lệ nhân viên
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Biểu đồ số lượng nhân viên
sns.countplot(
    data=df,
    x=TARGET,
    order=["No", "Yes"],
    ax=axes[0]
)

axes[0].set_title("Employee Attrition Count")
axes[0].set_xlabel("Attrition")
axes[0].set_ylabel("Number of Employees")

# Thêm số lượng lên đầu mỗi cột
for container in axes[0].containers:
    axes[0].bar_label(container)

# Biểu đồ tỷ lệ nhân viên
sns.barplot(
    x=attrition_percentage.index,
    y=attrition_percentage.values,
    order=["No", "Yes"],
    ax=axes[1]
)

axes[1].set_title("Employee Attrition Percentage")
axes[1].set_xlabel("Attrition")
axes[1].set_ylabel("Percentage (%)")

# Thêm tỷ lệ phần trăm lên đầu mỗi cột
for container in axes[1].containers:
    axes[1].bar_label(
        container,
        fmt="%.2f%%"
    )

plt.tight_layout()
plt.show()

# ============================================================
# CELL 6: ATTRITION BY GENDER
# Mục đích: Tìm hiểu sự khác biệt về nghỉ việc giữa nam và nữ
# ============================================================

# Tạo bảng đếm số nhân viên theo giới tính và tình trạng nghỉ việc
gender_count = pd.crosstab(
    df["Gender"],
    df[TARGET]
)

print("Employee count by Gender:")
display(gender_count)

# Tính tỷ lệ nghỉ việc trong từng nhóm giới tính
# normalize="index" giúp tính tỷ lệ trên tổng số nhân viên
# của từng nhóm giới tính
gender_rate = pd.crosstab(
    df["Gender"],
    df[TARGET],
    normalize="index"
) * 100

print("\nAttrition percentage by Gender:")
display(gender_rate.round(2))

# Vẽ biểu đồ số lượng nhân viên
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Gender",
    hue=TARGET
)

plt.title("Employee Attrition by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()

# Vẽ biểu đồ tỷ lệ nghỉ việc theo giới tính
gender_rate["Yes"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Attrition Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()

# ============================================================
# CELL 7: ATTRITION BY AGE
# Mục đích: Khám phá mối liên hệ giữa độ tuổi và nghỉ việc
# ============================================================

# Tính tuổi trung bình của hai nhóm nhân viên
age_statistics = df.groupby(TARGET)["Age"].agg(
    ["count", "mean", "median", "min", "max"]
)

print("Age statistics by Attrition:")
display(age_statistics.round(2))

# Vẽ histogram thể hiện phân phối độ tuổi
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="Age",
    hue=TARGET,
    bins=20,
    multiple="layer",
    alpha=0.5
)

plt.title("Age Distribution by Attrition")
plt.xlabel("Age")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()

# Vẽ boxplot để so sánh độ tuổi giữa hai nhóm
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x=TARGET,
    y="Age"
)

plt.title("Age Comparison by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Age")

plt.tight_layout()
plt.show()


# ============================================================
# CELL 8: ATTRITION BY DEPARTMENT
# Mục đích: Xác định sự khác biệt giữa các phòng ban
# ============================================================

# Đếm số nhân viên nghỉ việc và ở lại theo từng phòng ban
department_count = pd.crosstab(
    df["Department"],
    df[TARGET]
)

print("Employee count by Department:")
display(department_count)

# Tính tỷ lệ nghỉ việc trong từng phòng ban
department_rate = pd.crosstab(
    df["Department"],
    df[TARGET],
    normalize="index"
) * 100

print("\nAttrition rate by Department:")
display(department_rate.round(2))

# Vẽ biểu đồ số lượng nhân viên
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x="Department",
    hue=TARGET
)

plt.title("Employee Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

# Vẽ biểu đồ tỷ lệ nghỉ việc
department_rate["Yes"].sort_values(
    ascending=False
).plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")

plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

# ============================================================
# CELL 9: ATTRITION BY JOB SATISFACTION
# Mục đích: Phân tích mức độ hài lòng công việc và nghỉ việc
# ============================================================

# JobSatisfaction có giá trị từ 1 đến 4
# 1: Low
# 2: Medium
# 3: High
# 4: Very High

# Đếm số nhân viên theo mức độ hài lòng và tình trạng nghỉ việc
satisfaction_count = pd.crosstab(
    df["JobSatisfaction"],
    df[TARGET]
)

print("Employee count by Job Satisfaction:")
display(satisfaction_count)

# Tính tỷ lệ nghỉ việc theo từng mức độ hài lòng
satisfaction_rate = pd.crosstab(
    df["JobSatisfaction"],
    df[TARGET],
    normalize="index"
) * 100

print("\nAttrition rate by Job Satisfaction:")
display(satisfaction_rate.round(2))

# Vẽ biểu đồ số lượng nhân viên
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="JobSatisfaction",
    hue=TARGET
)

plt.title("Employee Attrition by Job Satisfaction")
plt.xlabel("Job Satisfaction")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()

# Vẽ biểu đồ tỷ lệ nghỉ việc
satisfaction_rate["Yes"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Attrition Rate by Job Satisfaction")
plt.xlabel("Job Satisfaction (1 = Low, 4 = Very High)")
plt.ylabel("Attrition Rate (%)")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# ============================================================
# CELL 10: ATTRITION BY OVERTIME
# Mục đích: Kiểm tra sự khác biệt về nghỉ việc giữa nhóm
# nhân viên có làm thêm giờ và không làm thêm giờ
# ============================================================

# Đếm số lượng nhân viên theo tình trạng làm thêm giờ
overtime_count = pd.crosstab(
    df["OverTime"],
    df[TARGET]
)

print("Employee count by OverTime:")
display(overtime_count)

# Tính tỷ lệ nghỉ việc trong từng nhóm làm thêm giờ
overtime_rate = pd.crosstab(
    df["OverTime"],
    df[TARGET],
    normalize="index"
) * 100

print("\nAttrition rate by OverTime:")
display(overtime_rate.round(2))

# Vẽ biểu đồ số lượng
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="OverTime",
    hue=TARGET
)

plt.title("Employee Attrition by OverTime")
plt.xlabel("OverTime")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()

# Vẽ biểu đồ tỷ lệ nghỉ việc
overtime_rate["Yes"].plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Attrition Rate by OverTime")
plt.xlabel("OverTime")
plt.ylabel("Attrition Rate (%)")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# ============================================================
# CELL 11: ATTRITION BY MONTHLY INCOME
# Mục đích: So sánh thu nhập giữa nhân viên nghỉ việc
# và nhân viên tiếp tục làm việc
# ============================================================

# Tính các chỉ số thống kê thu nhập theo tình trạng nghỉ việc
income_statistics = df.groupby(TARGET)["MonthlyIncome"].agg(
    ["count", "mean", "median", "min", "max", "std"]
)

print("Monthly Income Statistics:")
display(income_statistics.round(2))

# Vẽ boxplot so sánh thu nhập
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x=TARGET,
    y="MonthlyIncome"
)

plt.title("Monthly Income by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Monthly Income")

plt.tight_layout()
plt.show()

# Vẽ biểu đồ phân phối thu nhập
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="MonthlyIncome",
    hue=TARGET,
    bins=30,
    multiple="layer",
    alpha=0.5
)

plt.title("Monthly Income Distribution by Attrition")
plt.xlabel("Monthly Income")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()

# ============================================================
# CELL 12: ATTRITION BY YEARS AT COMPANY
# Mục đích: Kiểm tra thời gian làm việc tại công ty
# có liên quan như thế nào đến tình trạng nghỉ việc
# ============================================================

# Tính thống kê số năm làm việc theo tình trạng nghỉ việc
tenure_statistics = df.groupby(TARGET)["YearsAtCompany"].agg(
    ["count", "mean", "median", "min", "max", "std"]
)

print("Years at Company Statistics:")
display(tenure_statistics.round(2))

# Vẽ boxplot so sánh số năm làm việc
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x=TARGET,
    y="YearsAtCompany"
)

plt.title("Years at Company by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Years at Company")

plt.tight_layout()
plt.show()

# Vẽ histogram phân phối thâm niên
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="YearsAtCompany",
    hue=TARGET,
    bins=20,
    multiple="layer",
    alpha=0.5
)

plt.title("Years at Company Distribution by Attrition")
plt.xlabel("Years at Company")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()

# ============================================================
# CELL 13: ATTRITION BY DISTANCE FROM HOME
# Mục đích: Khám phá khoảng cách đi làm có liên quan
# đến tình trạng nghỉ việc hay không
# ============================================================

# Tính thống kê khoảng cách theo tình trạng nghỉ việc
distance_statistics = df.groupby(TARGET)["DistanceFromHome"].agg(
    ["count", "mean", "median", "min", "max"]
)

print("Distance From Home Statistics:")
display(distance_statistics.round(2))

# Vẽ boxplot so sánh khoảng cách
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x=TARGET,
    y="DistanceFromHome"
)

plt.title("Distance From Home by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Distance From Home")

plt.tight_layout()
plt.show()

# ============================================================
# CELL 14: CORRELATION MATRIX
# Mục đích: Khám phá mức độ tương quan tuyến tính giữa
# các biến số trong dataset
# ============================================================

# Chỉ lấy những cột có kiểu dữ liệu số
numeric_df = df.select_dtypes(
    include=["number"]
)

# Tính ma trận tương quan Pearson
correlation_matrix = numeric_df.corr()

# In ma trận tương quan
display(correlation_matrix.round(2))

# Vẽ biểu đồ heatmap
plt.figure(figsize=(17, 13))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0,
    linewidths=0.3,
    annot=False
)

plt.title("Correlation Matrix of Numerical Features")

plt.tight_layout()
plt.show()

# ============================================================
# CELL 15: ATTRITION RATE SUMMARY
# Mục đích: Tổng hợp tỷ lệ nghỉ việc theo nhiều đặc trưng
# ============================================================

# Danh sách các đặc trưng phân loại cần phân tích
group_columns = [
    "Gender",
    "Department",
    "OverTime",
    "MaritalStatus",
    "BusinessTravel",
    "JobRole"
]

# Tạo danh sách để lưu kết quả
summary_results = []

# Lặp qua từng đặc trưng
for column in group_columns:

    # Tính tỷ lệ nghỉ việc theo từng nhóm
    rate = (
        df.groupby(column)[TARGET]
        .apply(lambda x: (x == "Yes").mean() * 100)
    )

    # Đưa kết quả vào danh sách
    for group_name, percentage in rate.items():

        summary_results.append({
            "Feature": column,
            "Group": group_name,
            "Attrition Rate (%)": round(percentage, 2)
        })

# Chuyển danh sách kết quả thành DataFrame
attrition_summary = pd.DataFrame(summary_results)

# Sắp xếp theo tỷ lệ nghỉ việc giảm dần
attrition_summary = attrition_summary.sort_values(
    by="Attrition Rate (%)",
    ascending=False
)

# Hiển thị bảng tổng hợp
display(attrition_summary)

# ============================================================
# CELL 16: EDA SUMMARY
# Mục đích: Tổng hợp các thông tin quan trọng sau EDA
# ============================================================

# Tính tỷ lệ nghỉ việc chung
overall_attrition_rate = (
    (df[TARGET] == "Yes").mean() * 100
)

# Tính số lượng nhân viên nghỉ việc
total_attrition = (df[TARGET] == "Yes").sum()

# Tính số lượng nhân viên ở lại
total_not_attrition = (df[TARGET] == "No").sum()

# Hiển thị kết quả tổng quan
print("=" * 55)
print("EMPLOYEE ATTRITION - EDA SUMMARY")
print("=" * 55)

print(f"Total employees: {len(df)}")

print(f"Employees who left: {total_attrition}")

print(f"Employees who stayed: {total_not_attrition}")

print(f"Overall attrition rate: {overall_attrition_rate:.2f}%")

# Hiển thị tỷ lệ nghỉ việc theo một số đặc trưng
print("\nAttrition rate by OverTime:")
display(overtime_rate.round(2))

print("\nAttrition rate by Department:")
display(department_rate.round(2))

print("\nAttrition rate by Gender:")
display(gender_rate.round(2))

print("=" * 55)
```
